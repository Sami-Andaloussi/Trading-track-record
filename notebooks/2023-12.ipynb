# December 2023

### Short AUD/CHF into the RBA decision — sized wrong

***AUD/CHF** · short · closed · -0.75R · Dec 04, 2023*

Divergent monetary policy outlook between Switzerland (inflation at 1.4%) and Australia (5.4%) was the core idea here. I put this on before the RBA decision, a risk I was aware of at the time. I'd also been watching the AU10Y-CH10Y yield spread against AUD/CHF — a correlation that used to be tight but had decoupled since mid-2022, more so since summer: the spread was pushing higher while the pair just sat there, which is what got me interested in the first place.

Technically it wasn't as clean as I'd like: a short-term trendline that was messy (a wick through it followed by an ambiguous return, so I drew a second dotted trendline just to help visualize the structure), and the main resistance above hadn't actually broken yet — a red flag I noted myself at the time. There was a correction toward the 50% fib where I'd have liked to get in, but I missed that window (might have caught something similar in November, not sure). On a lower timeframe an expanding wedge rejected price and broke a high — the setup that would have let me get long there instead.

Instead of continuing higher, price reversed straight through my level, triggered my short, and hit my stop almost immediately — a direct consequence of putting the trade on ahead of the RBA decision. Price kept falling after that, so the underlying direction was actually right, I just wasn't in it anymore by then.

86% of retail was also long the pair — same side as me — a signal I should have paid a lot more attention to instead of just noting it and moving on. I sized this at 75% of my usual risk rather than full size because of that positioning, but looking back I should have gone to 50%, not 75% — there was no real justification for the higher number beyond wanting to make some money early in the month, coming off a rough November, and knowing I only had about two weeks of trading left before traveling. I don't regret the stop getting hit — that was a known, accepted risk — but the sizing on this one was sloppy.

### Long AUD/CAD via a EUR cross-pair realization — best trade of the month

***AUD/CAD** · long · closed · +1.90R · Dec 06, 2023*

I wanted to sell EUR/AUD (a view I'd already run profitably the month before) and, separately, had a technical buy setup on EUR/CAD. Realized that selling EUR/AUD and buying EUR/CAD in equal size is functionally the same trade as just buying AUD/CAD directly — so that's the pair I traded, especially since a buy setup was shaping up on AUD/CAD itself too, confirming the same view from a different angle.

Fundamentally, the AU10Y-CA10Y yield spread had just spiked higher while AUD/CAD, despite breaking its downtrend and starting to turn up, hadn't moved anywhere near as much — a gap I read as more reason to buy. I was also staying structurally bullish AUD on inflation, the same thesis I'd run on EUR/AUD in November. On top of that, 80% of retail was positioned short the pair while I wanted to buy — an extra green light on the contrarian side.

Technically: the daily downtrend had just broken with early signs of a bullish wave structure. A fib retracement of that impulse brought price back to the 50% level, which lined up with an old resistance. On the 2H, that same level had turned into support, reinforced by a 5-touch trendline — the 5th touch rejecting price again. Rather than buy straight into what looked like a sharp drop (even though it was holding the 50-SMA), I waited for confirmation on a 15-minute chart — an impulse breaking a short-term high — then fibbed that move and waited for a pullback before executing, a method I use a lot.

On sizing I split the entry across two stops: a wider one placed above a swing low rather than just beneath it, and an even wider second one. The idea was mostly fundamental, so even a retest of the 61.8% fib wouldn't have invalidated the thesis — hence wanting room. But one stop that wide on the full size wouldn't have given a decent risk/reward, so I split the position across the two instead.

Price went my way at first, then rolled back over and started to look like it was breaking down — annoying timing, right around when I was leaving Panama for Europe, with a tight stop just above the 61.8% that nearly got clipped by a weekend wick. To widen that stop to my second level without increasing my dollar risk, I closed a small piece of the position, which cost me -0.08R at that exact moment — a risk-management close, not a profit-take.

Price then turned back in my favor toward resistance, where I closed 50% — partly because of the holidays (less screen time, wanted to cut risk) and partly because I had another position open that was heavily correlated (NZD/CAD, also long, very correlated through CAD exposure). Price came back to retest the broken trendline and the 61.8% level; I wanted to buy back the piece I'd just closed but got distracted by personal stuff (car shopping for my mother) and never did. Outside of the holidays I probably would have re-entered, and it likely would have worked.

Price then resumed higher as expected. Closed another 25% there, and after the quiet Christmas stretch, price came back to resistance one more time and I closed the final 25%. Four exits total counting the risk-management partial at -0.08R. Not a spectacular profit in isolation, but my best trade of the month.

### A EUR/AUD short I didn't want to take, but couldn't ignore

***EUR/AUD** · short · closed · +0.28R · Dec 07, 2023*

I didn't really want to re-trade EUR/AUD at this point — I already had AUD exposure through NZD/CAD (long) and AUD/CAD (long, entered the day before), and I think I still had AUD/CHF open too. But the setup looked too clean to pass on.

Same bearish thesis I'd been running the month before — I was waiting on a confirmed break and got it. Technically, price pulled back to fib levels in what looked like a retest before resuming the fundamental downtrend; on a shorter timeframe I had a really nice 5-touch bearish trendline with the 50-SMA lining up almost exactly on the 5th touch — a setup I genuinely can't explain missing earlier. I waited for a countertrend line to break, then for a pullback to the fib levels to find a clean entry and stop.

The plan was to ride this one on the daily via a 1H execution, trailing the stop through the holidays to try to catch some momentum. Looking back, that was a slightly dumb idea on the timing front — that approach makes a lot more sense outside the holiday stretch, when thin liquidity doesn't cap how much momentum can actually build. It wouldn't have changed my entry, but it should have changed how I took profit — more incrementally, off support levels, rather than swinging for a full daily-scale move.

Entered in two pieces: first a limit sell at 0.5R (not full size, given how much AUD exposure I already had and the fact I hadn't wanted this trade in the first place). Price then tested a trendline, broke a new countertrend line, and I filled the rest to get back to full risk, this time with a tighter stop off the new trendline level. Retail was mostly long the pair too, the opposite side from me — another confluence.

Price kept dropping toward my first target; with hindsight, given how close Christmas was and how thin conditions were, I should have banked 50% there instead of holding out for a full daily-scale move I couldn't reasonably expect to play out over the holidays.

Support held, the trendline was respected, price dropped again then came back to resistance — I decided a break of a 7-touch trendline (visible to everyone, so likely to pull in volume) would be my exit signal. As soon as it broke, I closed both pieces: the first leg for +0.34R, the add-on for a small -0.06R, net +0.28R. Price kept falling afterward, confirming the fundamental thesis, but on ugly, flat price action I don't like sitting through — so I don't regret the early exit even though it capped the gain.
