# July 2023

### CAD/CHF short into the BoC decision — textbook buy-the-rumor-sell-the-fact

***CADCHF** · short · closed · +3.90R · Jul 01, 2023*

Canadian inflation came in cooler than expected — a good surprise that reduced the odds of further BoC tightening. When a central bank stops hiking while others keep going, capital tends to rotate out of that currency into higher-yielding ones, which was another reason to be short CAD.

On the 4H, contextualized by a daily downtrend, price came back to a rejected trendline that also lined up with a resistance level, and a counter-trendline was forming with an early bearish divergence. Retail was 76% long the pair — the contrarian signal I look for.

I don't like entering on a bare retest without a confirmed break, so on the initial retest of the counter-trendline I only took a half position, leaving a resting sell order for the other half at the 50% fib to fill if price broke down and corrected back to that level. That's exactly what happened: price broke the last high, corrected to the 61.8%, then filled my limit sell at the 50% fib for the second half — a slightly worse risk/reward than the first half, but a very good blended average overall.

Price kept falling without giving me the small pullback I was watching for to add, so I just kept an eye on a major support level below. The real reason I closed in two tranches was calendar-driven: the following week had the BoC rate decision, and with inflation already surprising lower the week before, the market had likely already priced in the end of hikes — classic buy-the-rumor-sell-the-fact dynamics. So I closed half the position once, then the rest shortly after the announcement.

Price kept falling after I was out, contrary to my expectation of a bounce around the announcement — but I'm completely happy with how I managed this one. If I had to redo it, I'd do it exactly the same way; it was the most logical, safest way to play it.

### Right process, wrong outcome on AUD/JPY long

***AUDJPY** · long · closed · -0.75R · Jul 06, 2023*

I wasn't running full risk this month, which is why this one cost -0.75R rather than a full -1R. The BoJ had refused to shift policy despite an improving economic backdrop and a pickup in underlying inflation, keeping its ultra-dovish language intact. The market was betting the status quo would hold a while longer, which should keep pushing the yen to multi-year lows against its major peers.

On a timeframe above the 2H chart I was watching, the trend was up: price had just broken a major resistance level, retested it as support, then broken higher again including the last high. I was looking for a pullback into fib levels to catch a small continuation move. Retail was 88% short — the contrarian read I wanted.

I got filled at the 61.8% fib rather than the 50% I used on the CAD/CHF trade, because the risk/reward at 50% wasn't good enough for my plan — not certain I'd get filled at all, but worth it if I did.

Price just didn't pump the way I needed it to and came back to stop me out at -0.75R. I separate a bad position — the decision process — from a bad outcome; a good decision can still lose, and vice versa. This was a good decision I'd take again exactly the same way. The point is to judge the process, not just the result.

### Faded an overdone dollar selloff on USD/JPY

***USDJPY** · long · closed · +1.98R · Jul 14, 2023*

The dollar had dropped 4% against the yen in a single week following US CPI. I compared the Dollar Index against 10-year Treasury yields, two things that are normally correlated — but that relationship had broken down since the start of 2023: the dollar had fallen hard without yields falling nearly as much, which told me the dollar was undervalued. The soft CPI print made the market think the Fed's hiking cycle was basically over, a read I partly agreed with, but not the size of the reaction — if money is leaving the dollar it has to go somewhere, and the only economy with more attractive rates than the US at the time was New Zealand, which had already stopped hiking itself. So nothing really justified a 4% drop in a single week. Either US yields were too high or the dollar was too cheap, and I started looking for dollar-long setups — like finding a brand-new laptop on sale for way less than it's worth.

Technically, price returned to a key level after that 4% drop, with an early bullish divergence forming. Same logic as the CAD/CHF trade earlier in the month: I waited for a break of the last highs — sellers exhausted, buyers back in control — before looking for a pullback into fib levels for entry. That break confirmed, followed by a pullback to the 50% fib (the 61.8% wasn't needed for a good risk/reward here).

I managed this one out in stages. First partial, about a third of the position, came at the first new highs, since a lot of traders were still targeting further dollar weakness and the rally was choppy — impulse, correction, repeat. Price then ran hard higher on a Friday right into the weekend close, a move that didn't feel especially logical to me at the time, so I took another partial there. The following Monday, price corrected on the open with FOMC coming up; I tightened my stop, and a wick took out the final third.

That last exit wasn't my best — I probably should have taken the remaining third off a bit earlier rather than letting the stop get tagged. Still, the net result landed around +1.98R.
