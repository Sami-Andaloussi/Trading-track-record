# August 2024

### Long EUR/JPY into the BOJ carry-trade unwind

***EURJPY** · long · closed · +0.56R · 2024-08*

August opened with a genuine shock: the BOJ hiked rates on July 31st, completely unexpected by most of the market, and it tore through the yen carry trade that had been running for months. EUR/JPY, which had a solid uptrend built on the ECB being hawkish against an ultra-accommodative BOJ, dropped something like 10% in a matter of days.

My read going into this trade was that the sell-off had gone further than the underlying policy shift justified. The bounce that followed felt timid relative to the size of the drop, but the BOJ's deputy governor came out afterward with more reassuring, less hawkish comments, which took some of the panic out of the market. Retail positioning was also telling me something useful — the crowd had flipped to net short on the pair, which is the opposite of what you'd normally expect (I'd usually see something like 95% long in a pair like this). That kind of positioning flip after a violent move is exactly the setup I look for. I also had a bullish engulfing candle on the daily to trigger the entry.

Given it was the middle of the summer holiday period and volatility (VIX) was elevated, I sized this at 60% of my normal risk and used a wider stop than usual to give it room. I trailed the stop as the trade worked and got taken out for +0.56R — which, at 60% risk, was effectively a full-risk win. Good trade to have on when everyone else was still shaken from the carry unwind.

### Long AUD/NZD on the RBA/RBNZ policy gap — stopped out

***AUDNZD** · long · closed · -1.00R · Aug 20, 2024*

The thesis here was a monetary policy divergence trade: the RBA was still holding a hawkish line, with core inflation running at 3.9% in Q2 2024, while the RBNZ had just delivered its first rate cut with core inflation down at 2.8%. That gap felt wide enough to lean on.

I built the position in two tranches — an initial 50%, then added the other 50% after price confirmed a bounce off a trendline. In hindsight, I ended up running 100% of my normal risk on this one rather than the reduced size I should have used given I was on vacation that week — the same discipline I applied to the EUR/JPY trade (60% risk) didn't carry over here, and that's on me.

The pair reversed on me and I got stopped out for -1R. I didn't drop the thesis though — I came back into AUD/NZD long again the following month on the same divergence story, just sized down after this result. Lesson noted: reduce size when I'm not fully at my desk to manage a trade, regardless of how good the fundamental story looks.
