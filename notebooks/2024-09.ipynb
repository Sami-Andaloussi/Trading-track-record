### Back into AUD/NZD long on the same rate-divergence idea

***AUDNZD** · long · still open · +0.40R floating · 2024-09*

This is a continuation of the AUD/NZD divergence idea I had going in August — the RBA holding a hawkish line while the RBNZ was leaning more dovish. That first attempt got stopped out for -1R when the pair reversed on me. I've come back into the same trade this month, still betting on the policy gap between the two central banks eventually showing up in the cross rate.

No fireworks yet — it's sitting on a small floating profit (~+0.4R) as of writing this. I haven't detailed my exact entry level or stop here, but the plan is the same as last time, just sized a bit more conservatively given how quickly the first attempt failed. Sometimes the right response to getting stopped out on a good thesis is just to size down and try again rather than abandon the idea.

### Passed on a fourth CAD/CHF short at the same level

***CADCHF** · passed on · 2024-09*

This would have been my fourth setup on CAD/CHF, and I ended up deciding not to take it. Price was back at the upper edge of a channel that's already been tested multiple times, and the CAD index was sitting on a support level dating back to 2022 while CHF was still in its own uptrend, respecting its trendline. Retail positioning was heavily skewed long on the pair (around 87%), which is normally exactly the kind of contrarian setup I like to fade.

But the price action itself didn't confirm it — I wasn't seeing the seller pressure I'd want to see at that level before committing risk. Rather than force a fourth attempt on the same idea just because the fundamentals and positioning still lined up, I stood aside. In the end the market didn't break the way the setup would have needed, so sitting this one out was the right call — a useful reminder that fundamentals and positioning can point one way while price action tells you to wait.

### Long GBP/AUD stopped out when the Fed's 50bp cut hit

***GBP/AUD** · long · closed · -0.75R · 2024-09*

I'd been tracking a composite strength index across the GBP majors — GBP/USD, GBP/JPY, EUR/GBP, GBP/CHF, GBP/CAD, GBP/AUD, GBP/NZD — and it was showing broad, consistent demand for sterling, with repeated rejections at the same resistance area over a couple of weeks. That was enough for me to want to be long GBP against something, and I picked AUD as the funding leg given its relative weakness at the time.

The trade didn't survive contact with the macro calendar. The Fed cut rates by 50 basis points that month, a bigger cut than a lot of people were positioned for, and it reignited a broad risk-on bid that pushed AUD strength right through my GBP thesis. I got stopped out for -0.75R.

The lesson I took from this one: a strong single-currency index is a good starting point, but it doesn't override what a specific cross does into a major central bank surprise. If I've got a big data print or rate decision sitting between me and my target, I need to weight that risk more heavily before sizing a trade off a broader index reading alone.
